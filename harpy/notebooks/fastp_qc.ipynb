```python
import altair as alt
import json
import os
import sys
import time
import polars as pl

from harpy.report.html import StatsBox
from harpy.report.plots import SafeRender
from harpy.report.tables import ITable
from harpy.report.utilities import FastpResults
```

```python
indir = "reports/data/fastp"
```

```python
fp = FastpResults(indir)
```

```python
(
    StatsBox()
    .add(fp.samples, 'Samples')
    .add(fp.max_len, 'Max R1 len before')
    .add(fp.max_len_after, 'Max R1 len after')
    .add(fp.max_len_r2, 'Max R2 len before')
    .add(fp.max_len_r2_after, 'Max R2 len after')
).render()
```

## Trimming Metrics

```python
_tables = os.path.join(os.path.dirname(os.path.dirname(indir)), "tables")
os.makedirs(_tables, exist_ok=True)
_ = fp.stats.write_csv(os.path.join(_tables ,'trim.stats.csv'))
ITable(fp.stats, "trim.stats.csv", fixedcols=1).render()
```

## Base Quality Scores
These reflect the mean base quality scores per sample at each base-pair position before and after filtering. Base quality scores are interpreted from the Phred scores. Each line corresponds to the average within a sample.

### Read 1

```python
qual_r1 = (
    alt.Chart(fp.qual_curves, width = 310, height = 300)
    .transform_window(index='row_number()')
    .transform_fold(list(fp.qual_curves.columns), as_ = ['Sample', 'Quality'])
    .mark_line()
    .encode(
        x=alt.X('index:Q', title = "Base Position"),
        y='Quality:Q',
        color=alt.Color("Sample:N", legend = None).scale(scheme='turbo')
    )
    .properties(title=alt.Title('Read 1 Base Quality Before Filtering'))
)

qual_r1_after = (
    alt.Chart(fp.qual_curves_after, width = 325, height = 300)
    .transform_window(index='row_number()')
    .transform_fold(list(fp.qual_curves_after.columns), as_ = ['Sample', 'Quality'])
    .mark_line()
    .encode(
        x=alt.X('index:Q', title = "Base Position"),
        y='Quality:Q',
        color=alt.Color("Sample:N", legend = None).scale(scheme='turbo')
    )
    .properties(title=alt.Title('Read 1 Base Quality After Filtering'))
)

qual_r1 | qual_r1_after
```

### Read 2

```python
qual_r2 = (
    alt.Chart(fp.qual_curves_r2, width = 310, height = 300)
    .transform_window(index='row_number()')
    .transform_fold(list(fp.qual_curves_r2.columns), as_ = ['Sample', 'Quality'])
    .mark_line(strokeWidth = 1.5)
    .encode(
        x=alt.X('index:Q', title = "Base Position"),
        y='Quality:Q',
        color=alt.Color("Sample:N", legend = None).scale(scheme='turbo')
    )
    .properties(title=alt.Title('Read 2 Base Quality Before Filtering'))
)

qual_r2_after = (
    alt.Chart(fp.qual_curves_r2_after, width = 325, height = 300)
    .transform_window(index='row_number()')
    .transform_fold(list(fp.qual_curves_r2_after.columns), as_ = ['Sample', 'Quality'])
    .mark_line(strokeWidth = 1.5)
    .encode(
        x=alt.X('index:Q', title = "Base Position"),
        y='Quality:Q',
        color=alt.Color("Sample:N", legend = None).scale(scheme='turbo')
    )
    .properties(title=alt.Title('Read 2 Base Quality After Filtering'))
)

qual_r2 | qual_r2_after
```

## GC Content
This reflects the percent G or C bases at each position a long a read. Each line corresponds to the average within a sample.

### Read 1

```python
gc_r1 = (
    alt.Chart(fp.gc_curves, width = 310, height = 300)
    .transform_window(index='row_number()')
    .transform_fold(list(fp.qual_curves.columns), as_ = ['Sample', 'GC'])
    .mark_line(strokeWidth = 1.5)
    .encode(
        x=alt.X('index:Q', title = "Base Position"),
        y=alt.Y('GC:Q').axis(format = "%"),
        color=alt.Color("Sample:N", legend = None).scale(scheme='turbo')
    )
    .properties(title=alt.Title('Read 1 GC Content Before Filtering'))
)

gc_r1_after = (
    alt.Chart(fp.gc_curves_after, width = 325, height = 300)
    .transform_window(index='row_number()')
    .transform_fold(list(fp.qual_curves_after.columns), as_ = ['Sample', 'GC'])
    .mark_line(strokeWidth = 1.5)
    .encode(
        x=alt.X('index:Q', title = "Base Position"),
        y=alt.Y('GC:Q').axis(format = "%"),
        color=alt.Color("Sample:N", legend = None).scale(scheme='turbo')
    )
    .properties(title=alt.Title('Read 1 GC Content After Filtering'))
)

gc_r1 | gc_r1_after
```
